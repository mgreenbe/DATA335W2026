```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

# Buffon's needle

```python
ax = plt.gca()
ax.set_aspect("equal")
ax.set_xlim(-1, 2)
ax.plot([0, 0], [-1, 2], "k--")
ax.plot([1, 1], [-1, 2], "k--")

n = 30
k = 0
for _ in range(n):
    x0 = stats.uniform.rvs()
    y0 = stats.uniform.rvs()
    theta = 2 * np.pi * stats.uniform().rvs()
    x1 = x0 + np.cos(theta)
    y1 = y0 + np.sin(theta)
    if np.floor(x0) != np.floor(x1):
        k += 1
        color = "C1"
    else:
        color = "C0"
    # color = "C0" if np.floor(x0) == np.floor(x1) else "C1"
    ax.plot([x0, x1], [y0, y1], color=color)

p = k / n
ax.set_title(f"$\\pi \\approx {2/p:.4f}$")
plt.show()
```

```python
n = 100_000
k = 0
for _ in range(n):
    x0 = stats.uniform.rvs()
    theta = 2 * np.pi * stats.uniform().rvs()
    x1 = x0 + np.cos(theta)
    if np.floor(x0) != np.floor(x1):
        k += 1
p = k / n
2 / p
```

```python
n = 100_000
x0 = stats.uniform.rvs(size=n)
theta = 2 * np.pi * stats.uniform().rvs(size=n)
x1 = x0 + np.cos(theta)
mask = ~np.isclose(np.floor(x0), np.floor(x1))
p = mask.mean()
2 / p
```

```python
n = 1_000_000
x = stats.uniform.rvs(size=n)
y = 2 * np.sqrt(1 - x**2)
plt.hist(y, bins=100)
plt.title(f"y.mean() = {y.mean():.4f}")
plt.show()
```

```python
n = 1_000_000
theta0 = 2 * np.pi * stats.uniform.rvs(size=n)
theta1 = 2 * np.pi * stats.uniform.rvs(size=n)
y = np.sqrt(
    (np.cos(theta1) - np.cos(theta0)) ** 2 + (np.sin(theta1) - np.sin(theta0)) ** 2
)
plt.hist(y, bins=100)
plt.title(f"y.mean() = {y.mean():.4f}")
plt.show()
```

```python
n = 1_000_000
theta0 = 2 * np.pi * stats.uniform.rvs(size=n)
theta1 = 2 * np.pi * stats.uniform.rvs(size=n)
y = np.sqrt((np.cos(theta1) + 1) ** 2 + (np.sin(theta1)) ** 2)
plt.hist(y, bins=100)
plt.title(f"y.mean() = {y.mean():.4f}")
plt.show()
```
