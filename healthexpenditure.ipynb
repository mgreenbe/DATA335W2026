```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
health = pd.read_csv(
    "healthexpenditure.csv", usecols=["Country", "2005"], index_col="Country"
)
life = pd.read_csv(
    "lifeexpectancy.csv", usecols=["Country", "2005"], index_col="Country"
)
health = health.rename(columns={"2005": "health"})
life = life.rename(columns={"2005": "life"})
df = health.merge(life, how="left", left_index=True, right_index=True)
df
```

```python
df.plot.scatter("health", "life")

for i, row in df.iterrows():
    plt.annotate(i, (row.health, row.life + 0.15), fontsize=8)
```
