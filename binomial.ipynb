```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

### Central Limit Theorem

Suppose $X_0, X_1, \ldots$ are IID random variables with
$$
\mathbb{E}(X_i) = \mu
\qquad\text{and}\qquad
\mathbb{V}(X_i) = \sigma^2.
$$

Let
$$
\bar{X}_n = \frac1n\sum_{i < n}X_i.
$$

Then
$$
\mathbb{E}(\bar{X}_n) = \mu
\qquad\text{and}\qquad
\mathbb{V}(\bar{X}_n) = \frac{\sigma^2}n.
$$

#### Central Limit Theorem
For large $n$,
$$
\bar{X}_n\sim N\left(\mu, \frac{\sigma^2}{n}\right)\qquad\text{approximately}.
$$

### The normal approximation to the binomial distribution

Let $X_0,X_1,\ldots$ be independent draws from $\operatorname{Bernoulli}(p)$.

Then
$$
\mathbb{E}(X_i) = p
\qquad\text{and}\qquad
\mathbb{V}(X_i) = p(1 - p).
$$

By the central limit theorem
$$
\bar{X}_n \sim N\left(p, \frac{p(1 - p)}{n}\right)
\qquad\text{approximately}.
$$

Set
$$
Y_n = \sum_{i < n}X_i = n\bar{X}_n.
$$

Then
$$
Y_n\sim \operatorname{Binomial}(n, p).
$$
and
$$
Y_n \sim N\left(np, np(1 - p)\right)\qquad\text{approximately}.
$$


```python
n = 100
p = 0.2
y = np.arange(50)
plt.bar(y, stats.binom.pmf(y, n=n, p=p), width=0.75, label="Binomial PMF")
grid = np.linspace(0, 50, 1000)
plt.plot(
    grid,
    stats.norm.pdf(grid, loc=n * p, scale=np.sqrt(n * p * (1 - p))),
    "C1",
    lw=2,
    label="Normal PDF",
)
plt.legend()
plt.show()
```
