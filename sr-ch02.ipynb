### Bayesian Inference

> *In order to make good inference about what actually happened, it helps to consider everything that could have happened.*

A bag contains four marbles, each either blue, $B$, or white, $W$.

There are five possible **conjectures** about the proportion $\theta$ of blue marbles in the bag:
$$
\theta\in\left\{\tfrac04,\; \tfrac14,\; \tfrac24,\; \tfrac34,\; \tfrac44\right\}
$$

We draw a sequence of marbles from the bag, with replacement.

Let
$$
X_n = \begin{cases}
1&\text{if the $n$-th draw is blue,}\\
0&\text{if the $n$-th draw is white.}
\end{cases}
$$

Let
$$
Y_n = X_0 + X_1 + \cdots + X_n.
$$

How can we assign meaningful plausibilities to the conjectures having observed $Y_n$?

#### Bayesian inference principle, take 1

In the absence of prior information about $\theta$,
$$
P\left(\theta\mid Y_n\right) \propto
P\left(Y_n \mid \theta\right) = \binom{n}{Y_n}\theta^2(1 - \theta)
$$

```python
P = np.array([theta**2 * (1 - theta) for theta in [0, 0.25, 0.5, 0.75, 1]])
P / P.sum()
```
