Let $\pi_P$ and $\pi_V$ be the probabilities of infection given placebo and vaccine, respectively.

Vaccine efficacy ($\operatorname{VE}$) is defined by
$$
\operatorname{VE} = \frac{\pi_P - \pi_V}{\pi_P} = 1 - \frac{\pi_V}{\pi_P}
$$

Let $n_P$ and $n_V$ be the numbers of patients treated with placebo and vaccine, respectively.

Let $Y_P$ and $Y_V$ be the numbers of cases observed among patients treated with placebo and vaccine, respectively.

$$
Y_P\sim \operatorname{Bin}(n_P, \pi_P),\qquad
Y_V\sim \operatorname{Bin}(n_V, \pi_V)
$$

Define 
$$
\theta = \frac{\pi_V}{\pi_P + \pi_V}.
$$

Then
$$
\theta = \frac{1 - \operatorname{VE}}{2 - \operatorname{VE}}
$$
and
$$
\operatorname{VE} = \frac{2\theta - 1}{\theta - 1}.
$$

```python
theta = np.arange(0, 1, 0.01)
VE = (2 * theta - 1) / (theta - 1)
plt.plot(theta, VE)
plt.show()
```
