### Statistical Rethinking, Exercises 2H

Suppose there are two species of panda bear. Both are equally common in the wild and live
in the same places. They look exactly alike and eat the same food, and there is yet no genetic assay
capable of telling them apart. They differ however in their family sizes. Species A gives birth to twins 10% of the time, otherwise birthing a single infant. Species B births twins 20% of the time, otherwise birthing singleton infants. Assume these numbers are known with certainty, from many years of field research.

```python
rng = np.random.default_rng()
n = 1000000
is_species_A = np.zeros(n, dtype=bool)
first_birth_twins = np.zeros(n, dtype=bool)
second_birth_twins = np.zeros(n, dtype=bool)
for k in range(n):
    is_species_A[k] = rng.uniform() < 0.5
    if is_species_A[k]:
        first_birth_twins[k] = rng.uniform() < 0.1
        second_birth_twins[k] = rng.uniform() < 0.1
    else:
        first_birth_twins[k] = rng.uniform() < 0.2
        second_birth_twins[k] = rng.uniform() < 0.2
```

Since we have no prior knowledge of the species of the mom, we assign priors
$$
P(A)=P(B)=0.5.
$$

We compute the posterior probabilities for the pandas species conditioned on the first observed birth being twins.

$$
\begin{aligned}
P(A\mid\text{twins})
&= \frac{P(\text{twins}\mid A)P(A)}{P(\text{twins}\mid A)P(A) + P(\text{twins}\mid B)P(B)}\\
&= \frac{0.1\times 0.5}{0.1\times 0.5 + 0.2\times 0.5}\\
&= \frac13
\end{aligned}
$$

It follows that
$$
P(B\mid\text{twins}) = \frac23.
$$

Compute the probability that the second observed birth is also twins.

$$
\begin{aligned}
P(\text{twins}) &= P(\text{twins}\mid A)P(A) + P(\text{twins}\mid B)P(B)\\
&= 0.1\times\frac13 + 0.2\times\frac23\\
&= \frac16
\end{aligned}
$$

Compute the probability that the pandas is Species A, given that the second birth is also twins.

$$
\begin{aligned}
P(A\mid\text{twins})
&= \frac{P(\text{twins}\mid A)P(A)}{P(\text{twins})}\\
&= \frac{0.1\times \frac13}{\frac16}\\
&= \frac15
\end{aligned}
$$


```python
is_species_A[first_birth_twins].mean()
```
